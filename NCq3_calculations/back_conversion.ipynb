```python
import random
szavak = []
with open('szavak.txt', 'r') as file:
    for line in file:
        szavak += [line.strip()]
    interval = [i for i in range(50)]
    print(interval)
    random.shuffle(interval)
    mixed = [szavak[int(i)] for i in interval]
with open('szavak_out.txt', 'w') as file:
    for m in mixed:
        file.write(str(m) + '\n')
```
